```python
def limpeza(data):
    #1.  Columns lowercase
    data.columns = data.columns.str.lower()
    #2.  year_of_release as int
    #note that the lowercase has already been implemented
    data['year_of_release'] = pd.to_numeric(data['year_of_release'], errors='coerce').astype('Int64')
    #3. Mudar tbd do user_score para nulo e conversão para float
    data['user_score'] = pd.to_numeric(data['user_score'], errors='coerce')
    #and finally, the conversion
    return data

#I like this if condition a lot. It helps to avoid errors produced by the non-linear execution of the code.    
if 'Name' in games.columns:
    print('Run Once')
    games = limpeza(games)
else:
    print('Already Done')
```

## Data Analysis:

With the cleanup completed, it's time to look for patterns.
Without losing sight of our goal, let's start right away by investigating some data in search of clues that may lead us to patterns:

### Name:
Certain intellectual properties and their studios have large audiences of fans accompanying them. Games like *Call Of Duty* or *Fifa* have guaranteed sales because these games are popular. However, cases like these are a minority and represent peaks, not success patterns. Despite this, we all know that as soon as the next one releases, everyone will run to get (*coughing*, GTA 6, *coughing*) (funny enough I was right, GTA VI indeed released and it was a massive success lol)

### Platform:
Considering that this database relates to the year 2016, it is logical that video games coming from the **"titans"** (biggest consoles in the market) are more successful than portable consoles. But the fact that the competition is practically limited to PS4, Xbox One and Nintendo Switch, is the opposite case of the Name, in which virtually all sales are already within these categories and little information exists outside of them. One important point to make is that several new consoles were released in 2016: the PS4 Pro, PS4 Slim, Xbox One S, and PlayStation VR. This marks a key turning point where sales for newer hardware began to rise at the expense of older consoles.


### Year of Release
Over the years, there is a tendency to define another previous period as "retro". I believe that certain games, having been a part of many childhoods, come to a point where it becomes relevant again to a particular age group thanks to the effects of bitter-sweet, cozy and familiar, *Nostalgia*, and the better days.
In sociology and marketing, this concept is called the "20-year rule"** and as the name itself implies, a person feels a level of nostalgia elevated by a game he played 20 years ago**. In this case they would be games released in the years 90**. This in itself would formulate an excellent marketing campaign to revive sales of products from this decade. ** We have a strong point to explore here. **

###  Genres:
Statistically speaking, genres function as ice cream flavors. Of course anyone can buy any flavor, but when observing in the general context, **people tend to gravitate towards a particular gender/taste**. We have another strong point here.

### Sales USA / Sales Europe / Sales Japan / Sales Other Countries:
By using English, advertising campaigns can cover both the European and the American markets. As such, given the diversity of social classes and cultures, it is a more attractive market, but also more competitive. Sales in these two sectors are much more important. Together they represent 16.5% of the world market, while Japan only accounts for 1.5%.
If we look at this from a statistical scope, we can use **TAM (Total Addressable Market—or essentially, relying only on people/countries who have purchasing power to buy video games)** to calculate the real percentage of market presence. The numbers are a little different, putting Europe at 21.5%, the USA at 8.6%, and Japan at 2.2% of countries that are both internet users and part of the **"consumer class" that has enough money to spend on video games**, the type of market that an online video game store needs. This is important because of NTSC and PAL formats, which **are negligible factors in modern games, but very important in old/retro games.**
In short, if we want to maximize sales of retro games, **we have to promote PAL (Europe) versions more than the NTSC (USA) version because TAM (the target market) is bigger in Europe.**

### Critic Score / User Score:
According to statistical studies and calculations of Google Gemini, **the user score is mostly negligible**, affecting only in the long term (games that were released a few years ago) by sharing personal experiences. On the other hand, the real meat of these studies is in the critic score:
As critics receive the game before the release, they aren't as influenced by the hype of advertising campaigns as users and are generally more impartial in their analysis. **These previously mentioned statistics dictate that the critic score has to be 80 to be considered a success, and that every 5 points above that, its sales are designed to double!** That seems like a pattern to me! By comparison, scores of 70 or below run the risk of snowball, where the lower that number, the higher the sales penalty, which can only be saved through aggressive adjustment in the marketing budget. The video game **"Destiny"**, released in 2014 is a perfect example of this case.
Therefore, I conclude that close to 80% of my statistical conclusions on the next pages will be taken from the **critic score.**

### Rating:
Although **E** (All ages) have the highest **TAM**, the vast majority of profits come from **T** (Teen) and M (Mature). **T** in particular encapsulates children who want to feel older and adults who want to feel younger. Still, the highest-selling games (Call of Duty, GTA) are classified as **M**, this makes **E** the rating with most games, but **M** the rating with highest number of sales combined. Therefore, **T** looks to be the best of both worlds, at least at first glance.

Through the calculation of the t test, we obtain the p value. The p-value is a "trust measure" in which the higher the number is, the greater the confidence there is that the two t-test targets are equal. As in everything, there are false-positives, hence a margin of error is chosen to have a good combination of accuracy without causing too many false-positives, that margin is 0.05. In casual terms, having a chance to survive 95% is almost as good as 100%, isn't it? Because with alpha 0.05, only 1 in 20 cases becomes false positive. Statistically speaking, it's a good case.
